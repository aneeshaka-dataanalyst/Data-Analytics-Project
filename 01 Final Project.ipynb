## Introduction

Tuberculosis (TB) is a contagious infectious disease caused by Mycobacterium tuberculosis, primarily affecting the lungs, although it can also impact other parts of the body. Despite being preventable and curable, tuberculosis remains one of the leading causes of illness and death worldwide, particularly in low- and middle-income countries.

The purpose of this analysis is to understand the global burden of tuberculosis by examining key indicators such as TB incidence, mortality, treatment coverage, and treatment success rates, and to present meaningful insights through effective data visualization. This project uses Python libraries such as Pandas, NumPy, Plotly, Matplotlib, and Seaborn to analyze the tuberculosis burden dataset.

This analysis aims to understand global TB trends, identify high-burden regions and countries, and explore the relationships among key TB indicators.

## Data Source

The dataset used in this project was collected from the World Health Organization (WHO) Global Tuberculosis Programme, which provides official global tuberculosis statistics reported by member countries. The original dataset contains country-wise information on tuberculosis burden, including indicators such as TB incidence, mortality, population, case notifications, and treatment outcomes.

Note:- The original file name was "TB Burden Estimates by Country", but i renamed it to "Tuberculosis Burden Analysis". Only the file name was changed; the data structure and contents remain unchanged.

For the purpose of this project, the original dataset was renamed from "TB Burden Estimates by Country" to "Tuberculosis Burden Analysis". Additionally, a few columns that were not required for the analysis were removed. These include:

- ISO Country Code
- GBD Region Code
- Data Source Identifier
- Reporting Notes
- Reference URL

These columns were excluded because they primarily contained administrative identifiers and metadata that were not essential for the exploratory data analysis and visualizations. Removing them simplified the dataset while retaining all the relevant information required for meaningful analysis of tuberculosis trends, disease burden, and country-wise comparisons.

#### (iii) Bivariate Analysis

It examines the relationship between two variables to identify patterns, associations, and differences. It helps understand how one variable changes or relates to another within the tuberculosis burden dataset.

```python
# Aggregation Function to Summarize TB Burden Data
def TB_Stats(column):
    tb = df.groupby(column).agg(
        Total_Population=("Population", "sum"),
        Total_TB_Cases=("Estimated_TB_Cases", "sum"),
        Average_TB_Incidence=("TB_Incidence_Rate", "mean"),
        Total_TB_HIV_Cases=("Estimated_TB_HIV_Cases", "sum"),
        Average_TB_HIV_Percentage=("TB_HIV_Percentage", "mean"),
        Total_TB_Deaths=("Estimated_TB_Deaths", "sum"),
        Average_TB_Mortality=("TB_Mortality_Rate", "mean"),
        Average_Case_Fatality_Rate=("Case_Fatality_Rate", "mean"),
        Average_Case_Detection_Rate=("Case_Detection_Rate", "mean")
    ).round(2)

    return tb
```

***Info:-*** **Groups the data** by a selected column using the ***groupby()*** function and summarizes key tuberculosis indicators for each group.

```python
# TB Burden by WHO Region
TB_Stats("WHO_Region")
```

```python
# TB Burden by Country
TB_Stats("Country")
```

```python
# TB Burden by Year
TB_Stats("Year")
```

```python
# Top 10 Countries by Total TB Cases
df.groupby("Country").agg(
    Total_TB_Cases=("Estimated_TB_Cases","sum")
).sort_values(
    by="Total_TB_Cases",
    ascending=False
).head(10)
```

```python
# Top 10 Countries by Total TB Deaths
df.groupby("Country").agg(
    Total_TB_Deaths=("Estimated_TB_Deaths","sum")
).sort_values(
    by="Total_TB_Deaths",
    ascending=False
).head(10)
```

```python
# Top 10 Countries by Total TB-HIV Cases
df.groupby("Country").agg(
    Total_TB_HIV_Cases=("Estimated_TB_HIV_Cases","sum")
).sort_values(
    by="Total_TB_HIV_Cases",
    ascending=False
).head(10)
```

### (i) Total TB Cases by WHO Region

• ***Visualization Type:-*** Bar Chart (*Seaborn*)  
• ***Purpose:-*** To Compare the total estimated TB cases across WHO regions  
• ***Insight:-*** The chart shows that the ***South-East Asia (SEA)*** region has the *highest estimated TB cases*, followed by the ***Western Pacific (WPR)*** and ***African (AFR)*** regions, while the Region of the ***Americas (AMR)*** has the *lowest estimated TB cases*.

```python
# Import Required Libraries
import matplotlib.pyplot as plt
import seaborn as sns
# Calculate Total TB Cases by WHO Region
tb_cases = (
    df.groupby("WHO_Region")["Estimated_TB_Cases"]
      .sum()
      .sort_values(ascending=False)
      .reset_index()
)
# Define custom colors
colors = [
    "#4E79A7",  # Blue
    "#F28E2B",  # Orange
    "#E15759",  # Red
    "#76B7B2",  # Teal
    "#59A14F",  # Green
    "#EDC948"   # Yellow
]
# Create Figure
plt.figure(figsize=(10,6))
# Create Bar Chart
sns.barplot(
    data=tb_cases,
    x="WHO_Region",
    y="Estimated_TB_Cases",
    hue="WHO_Region",      # Assign different colors
    palette=colors,
    legend=False
)
# Chart Title
plt.title("Total TB Cases by WHO Region",fontsize=16,fontweight="bold")
# Axis Labels
plt.xlabel("WHO Region",fontsize=12)
plt.ylabel("Estimated TB Cases",fontsize=12)
# Rotate X-axis Labels
plt.xticks(rotation=20)
# Display Chart
plt.show()
```

### (ii) Share of Total TB Cases by WHO Region

• ***Visualization Type:-*** Pie Chart (*Matplotlib*)  
• ***Purpose:-*** To Show the percentage contribution of each WHO region to the global TB burden.  
• ***Insight:-*** The chart shows that the ***South-East Asia (SEA)*** region contributes the *largest share (35.7%)* of the total estimated TB cases, whereas the ***European (EUR)*** and Region of the ***Americas (AMR)*** contribute only a *small proportion*.

```python
# Calculate Total TB Cases by WHO Region
tb_cases = (
    df.groupby("WHO_Region")["Estimated_TB_Cases"]
      .sum()
)
# Custom Colors
colors = [
    "#4E79A7",  # Blue
    "#F28E2B",  # Orange
    "#E15759",  # Red
    "#76B7B2",  # Teal
    "#59A14F",  # Green
    "#EDC948"   # Yellow
]
# Create Figure
plt.figure(figsize=(8,8))
# Pie Chart
plt.pie(
    tb_cases,
    labels=tb_cases.index,
    colors=colors,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={"edgecolor":"white","linewidth":1.5},
    textprops={"fontsize":11}
)
# Chart Title
plt.title(
    "Share of Total Estimated TB Cases by WHO Region",
    fontsize=16,
    fontweight="bold"
)
# Equal Aspect Ratio
plt.axis("equal")
# Display Chart
plt.show()
```

### 🔎 Key Analytical Insights from Visual Exploration

#### (i) WHO Regional Differences in TB Burden

WHO regional analysis revealed considerable differences in the tuberculosis burden across the world. The African Region (AFR) consistently recorded the highest TB incidence, mortality, and TB-HIV burden, while the European Region (EUR) maintained comparatively lower disease burden and higher case detection rates. These findings highlight the unequal distribution of tuberculosis across different regions.

#### (ii) Country-wise TB Burden

Country-level analysis identified significant variation in tuberculosis burden among countries. India recorded the highest estimated number of TB deaths, while countries such as Djibouti, South Africa, Eswatini, and Lesotho reported the highest TB incidence rates. These countries represent priority areas for strengthened tuberculosis prevention and control efforts.

#### (iii) TB-HIV Co-infection Patterns

The analysis showed that TB-HIV co-infection remains a major public health concern, particularly in the African Region. Countries with higher TB-HIV cases also tended to experience higher TB incidence and mortality, highlighting the close relationship between HIV infection and tuberculosis burden.

#### (iv) Progress in Case Detection

Case detection rates showed a steady improvement over the study period across most WHO regions. Regions with higher case detection rates generally recorded better tuberculosis control outcomes, while regions with lower detection rates continued to experience a greater disease burden.

#### (v) Long-term Trends and Relationships

Trend analysis indicated that both TB incidence and mortality rates have gradually declined over the years, reflecting progress in global tuberculosis control. Correlation analysis also demonstrated strong positive relationships between TB incidence, estimated TB cases, and TB deaths, while improved case detection was associated with lower disease burden and better health outcomes.

### 📌 Overall Findings and Conclusion

Overall, this analysis demonstrates that the global tuberculosis burden is strongly influenced by regional disparities, disease incidence, mortality, and TB-HIV co-infection. Although significant progress has been made in reducing TB incidence and mortality over the years, several countries and WHO regions continue to experience a disproportionately high disease burden.

The findings emphasize that tuberculosis is shaped by the combined influence of multiple factors, including geographical location, healthcare access, case detection, and TB-HIV prevalence. These insights highlight the importance of strengthening early diagnosis, improving treatment coverage, and implementing targeted public health interventions in high-burden regions to further reduce the global impact of tuberculosis.

The findings can support evidence-based planning and policy decisions for tuberculosis prevention and control.

### 💡Recommendations

Based on the findings of this analysis, greater emphasis should be placed on strengthening tuberculosis prevention and control strategies in high-burden countries and WHO regions. Expanding early screening and improving access to timely diagnosis can help identify TB cases sooner and reduce disease transmission.

Healthcare systems should continue to improve treatment accessibility and encourage treatment adherence to reduce TB-related mortality and prevent the development of drug-resistant tuberculosis. Special attention should also be given to integrating TB and HIV healthcare services in regions with a high TB-HIV burden.

Increasing public awareness through education campaigns, improving healthcare infrastructure, and strengthening disease surveillance systems can further support tuberculosis control efforts. Continued monitoring of TB trends and targeted interventions in the most affected regions will contribute to reducing the global tuberculosis burden and support progress toward achieving the WHO End TB Strategy goals.

Future studies can include machine learning models to predict tuberculosis burden and identify high-risk countries.
